# Aquanga - 03. Feature Engineering

This notebook demonstrates the transformation of raw water quality records into machine learning feature matrices: CPCB compliance indicators, temporal lag features, rate-of-change trend velocity, and sequence creation.

```python
import os
import sys
import pandas as pd
import numpy as np

sys.path.insert(0, os.path.abspath(".."))
from ml.feature_engineering import generate_water_quality_features, generate_lag_features
from ml.create_sequences import get_chronological_split, fit_and_scale_data, reshape_for_sequences
```

## 1. Load Clean Dataset and Compute Compliance Flags

```python
clean_path = os.path.join("..", "data", "processed", "ganga_water_quality_clean.csv")
clean_df = pd.read_csv(clean_path)

df_flags = generate_water_quality_features(clean_df)
df_flags.head()
```

## 2. Generate Temporal Lags & Velocity Differences
To strictly avoid target leakage, rate of change is computed as `(lag1 - lag2)`.

```python
df_features = generate_lag_features(df_flags)
print(f"Feature matrix shape: {df_features.shape}")
df_features.head()
```

## 3. Chronological Train / Test Splitting

```python
X_train, X_test, y_train, y_test = get_chronological_split(df_features, test_year=2015)
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
```

## 4. Feature Scaling & 3D Sequence Construction

```python
X_train_s, X_test_s, y_train_s, y_test_s, sc_X, sc_y = fit_and_scale_data(X_train, X_test, y_train, y_test)
X_train_seq = reshape_for_sequences(X_train_s)
X_test_seq = reshape_for_sequences(X_test_s)

print(f"Tabular 2D X_train: {X_train_s.shape}")
print(f"Sequence 3D X_train for CNN/LSTM: {X_train_seq.shape}")
```
