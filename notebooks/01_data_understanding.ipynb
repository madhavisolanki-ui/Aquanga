# Aquanga - 01. Data Understanding

This notebook inspects the raw CPCB Ganga River water quality dataset (2011–2015), analyzes data structures, identifies missing values, and establishes the foundational understanding for the predictive monitoring system.

```python
import pandas as pd
import numpy as np
import os

raw_path = os.path.join("..", "data", "raw", "ganga_water_quality_2011_2015.csv")
df_raw = pd.read_csv(raw_path)
print(f"Raw dataset shape: {df_raw.shape}")
df_raw.head(10)
```

## 1. Column Inspection & Identification
Let's identify the location column, water-quality parameters (DO, BOD, Fecal Coliform), and years present in the dataset.

```python
print("Columns in raw dataset:")
for i, col in enumerate(df_raw.columns):
    print(f"{i+1}. {col}")
```

## 2. Missing Value & Data Type Analysis

```python
print("Missing values per column:")
print(df_raw.isna().sum())

# Check for non-numeric or string representations like 'NA'
print("\nSummary of textual 'NA' entries:")
for col in df_raw.columns:
    na_count = (df_raw[col].astype(str).str.strip().str.upper() == 'NA').sum()
    if na_count > 0:
        print(f"Column '{col}' contains {na_count} 'NA' tokens.")
```

## 3. Station Inventory

```python
stations = df_raw.iloc[:, 1].dropna().unique()
print(f"Total unique monitoring stations ({len(stations)}):")
for idx, st in enumerate(stations, 1):
    print(f"{idx}. {st}")
```
