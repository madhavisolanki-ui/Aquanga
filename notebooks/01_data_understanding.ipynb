```python
import pandas as pd

df = pd.read_csv("/content/ganga_water_quality_2011_2015.csv")

df.head()
```

```python
df.shape
```

```python
df.info()
df.isnull().sum()
```

```python
df = df.dropna(axis=1, how="all")
```

```python
df.shape
```

```python
print(df.columns.tolist())
```

```python
df.columns = [
    "sr_no",
    "location",
    "do_2011",
    "do_2012",
    "do_2013",
    "do_2014",
    "do_2015",
    "bod_2011",
    "bod_2012",
    "bod_2013",
    "bod_2014",
    "bod_2015",
    "fecal_coliform_2011",
    "fecal_coliform_2012",
    "fecal_coliform_2013",
    "fecal_coliform_2014",
    "fecal_coliform_2015"
]
```

```python
df = df.replace("NA", pd.NA)
```

```python
df.isnull().sum()
```

```python
numeric_columns = [
    "do_2011", "do_2012", "do_2013", "do_2014", "do_2015",
    "bod_2011", "bod_2012", "bod_2013", "bod_2014", "bod_2015",
    "fecal_coliform_2011", "fecal_coliform_2012",
    "fecal_coliform_2013", "fecal_coliform_2014",
    "fecal_coliform_2015"
]

df[numeric_columns] = df[numeric_columns].apply(
    pd.to_numeric, errors="coerce"
)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df["location"].unique()
```

```python
do_df = df.melt(
    id_vars=["sr_no", "location"],
    value_vars=["do_2011", "do_2012", "do_2013", "do_2014", "do_2015"],
    var_name="year",
    value_name="do"
)

do_df["year"] = do_df["year"].str.extract(r"(\d{4})").astype(int)
```

```python
bod_df = df.melt(
    id_vars=["sr_no", "location"],
    value_vars=["bod_2011", "bod_2012", "bod_2013", "bod_2014", "bod_2015"],
    var_name="year",
    value_name="bod"
)

bod_df["year"] = bod_df["year"].str.extract(r"(\d{4})").astype(int)
```

```python
fecal_df = df.melt(
    id_vars=["sr_no", "location"],
    value_vars=[
        "fecal_coliform_2011",
        "fecal_coliform_2012",
        "fecal_coliform_2013",
        "fecal_coliform_2014",
        "fecal_coliform_2015"
    ],
    var_name="year",
    value_name="fecal_coliform"
)

fecal_df["year"] = fecal_df["year"].str.extract(r"(\d{4})").astype(int)
```

```python
clean_df = do_df[["sr_no", "location", "year", "do"]].merge(
    bod_df[["sr_no", "year", "bod"]],
    on=["sr_no", "year"]
).merge(
    fecal_df[["sr_no", "year", "fecal_coliform"]],
    on=["sr_no", "year"]
)
```

```python
clean_df.to_csv(
    "/content/ganga_water_quality_clean.csv",
    index=False
)
```
