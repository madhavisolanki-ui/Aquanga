## 2. Model Performance Visualization

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
x = np.arange(len(results_df))
width = 0.35

ax.bar(x - width/2, results_df["MAE"], width, label="MAE (mg/L)", color="#3182ce")
ax.bar(x + width/2, results_df["RMSE"], width, label="RMSE (mg/L)", color="#dd6b20")

ax.set_title("Model Comparison: MAE and RMSE Metrics on 2015 Held-out Test Year", fontsize=11, fontweight="bold")
ax.set_xticks(x)
ax.set_xticklabels(results_df["Model"], rotation=30, ha="right")
ax.set_ylabel("Error (mg/L)")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.4, axis="y")
plt.tight_layout()
plt.show()
```
