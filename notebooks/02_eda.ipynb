# Aquanga - 02. Exploratory Data Analysis (EDA)

Comprehensive statistical analysis, distribution plots, temporal trends, and parameter correlation for the Ganga River water quality observations.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
clean_path = os.path.join("..", "data", "processed", "ganga_water_quality_clean.csv")
if not os.path.exists(clean_path):
    import sys
    sys.path.insert(0, os.path.abspath(".."))
    from ml.preprocessing import run_preprocessing
    df = run_preprocessing(output_path=clean_path)
else:
    df = pd.read_csv(clean_path)

df.head()
```

## 1. Summary Statistics

```python
df.describe()
```

## 2. Parameter Distributions
Visualize distributions of DO, BOD, and Fecal Coliform with CPCB standard threshold markers.

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))

# DO Distribution
sns.histplot(df["do"], kde=True, ax=axes[0], color="#3182ce")
axes[0].axvline(5.0, color="green", linestyle="--", label="CPCB Min (5.0 mg/L)")
axes[0].axvline(4.0, color="red", linestyle=":", label="CPCB Hypoxia (<4.0 mg/L)")
axes[0].set_title("Dissolved Oxygen (DO) Distribution")
axes[0].legend()

# BOD Distribution
sns.histplot(df["bod"], kde=True, ax=axes[1], color="#dd6b20")
axes[1].axvline(3.0, color="green", linestyle="--", label="CPCB Clean (<=3.0 mg/L)")
axes[1].axvline(6.0, color="red", linestyle=":", label="CPCB Severe (>6.0 mg/L)")
axes[1].set_title("Biochemical Oxygen Demand (BOD) Distribution")
axes[1].legend()

# Fecal Coliform Distribution (log scale)
sns.histplot(np.log10(df["fecal_coliform"]), kde=True, ax=axes[2], color="#805ad5")
axes[2].axvline(np.log10(2500), color="green", linestyle="--", label="CPCB Limit (2500 MPN)")
axes[2].set_title("Log10(Fecal Coliform) Distribution")
axes[2].legend()

plt.tight_layout()
plt.show()
```

## 3. Station-Wise Longitudinal Trends (2011–2015)

```python
plt.figure(figsize=(14, 6))
sns.lineplot(data=df, x="year", y="do", hue="location", marker="o", linewidth=2)
plt.axhline(5.0, color="green", linestyle="--", label="CPCB Threshold (>= 5 mg/L)")
plt.title("Dissolved Oxygen Trends across Ganga Monitoring Stations (2011-2015)", fontsize=13, fontweight="bold")
plt.xlabel("Year")
plt.ylabel("DO (mg/L)")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 4. Parameter Correlation Analysis

```python
plt.figure(figsize=(7, 5))
sns.heatmap(df[["do", "bod", "fecal_coliform"]].corr(), annot=True, cmap="Blues", fmt=".2f", cbar=True)
plt.title("Water Quality Parameter Correlation Matrix")
plt.tight_layout()
plt.show()
```
